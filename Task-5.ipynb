# Handling Missing Values/Null Values
## IMPUTATION METHOD
## Syntax:dataframe.fillna(value="any value",inplace=True)

```python
# Filling missing values in the age column
titanic_df['age'].fillna(value='male',inplace=True)
```

```python
# Filling the missing values in the embarked column
titanic_df['embarked'].fillna(value='S',inplace=True)
```

# value_counts
## Counts the number of times the each value occured in a dataframe
## Syntax:dataframe.value_counts()

```python
# finding the number the number of values occuring each time using value_counts
titanic_df['deck'].value_counts()
```

```python
# Filling the top value in deck column using above insight that value 'c' has high value
titanic_df['deck'].fillna(value='C',inplace=True)
```

```python
# Filling the null values in embark_town
titanic_df['embark_town'].fillna(value='Southampton',inplace=True)
```

## Note:In the above i have replaced with the specific values because these are the most or top occuring values so i used them to replace the null values

# astype
## used to convert the column from one data type to other.
## Converting age data type to int data type.
## Syntax:dataframe['column_name'].astype(required data type)
## Syntax:dataframe['column_name']=pd.to_numeric(dataframe['column_name'])


```python
# Converting 
titanic_df['age'] = pd.to_numeric(titanic_df['age'])
```

# NOTE:
## While converting to the data type of age column to int it throwing an error so replacing the values male with the mean of that column

```python
# Replacing the value male in column with mean of that column
titanic_df['age']=titanic_df['age'].replace('male',titanic_df['age'].mean)
```

```python
# Converting the age column to int data type
titanic_df['age'] = pd.to_numeric(titanic_df['age'],errors='coerce')
```

# Handling/Checking Outliers
## Outliers are the data points that does not belong to the group.
## Outlier could occur at both higher and lower end of the data.


```python
# Checking for outliers in age anf fare column
plt.boxplot(x=titanic_df['age'])
plt.title("Outliers of Age")
plt.show()
```

```python
# Checking for outliers in age anf fare column
plt.boxplot(x=titanic_df['fare'],vert=False)
plt.title("Outliers of fare")
plt.show()
```

# Note:In the above we can observe that there are no ouliers in age column but there are outliers in fare column.


```python
# Handling ouliers in fare column
Q1=titanic_df['fare'].quantile(0.25)
Q3=titanic_df['fare'].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outliers=titanic_df[(titanic_df['age']<lower_bound)|(titanic_df['fare']>upper_bound)]
print(outliers)
```

```python
# Including the ouliers
titanic_df['fare']=titanic_df['fare'].clip(lower=lower_bound,upper=upper_bound)
```

```python
# Checking for outliers of fare column after handling the outliers
plt.boxplot(x=titanic_df['fare'],vert=False)
plt.title("Outliers of fare")
plt.show()
```

# NOTE: In the above we can observe that the outliers are handled
